## Exercice 1 — Implémenter la sigmoid et sa dérivée
Écris la fonction sigmoid et sigmoid_grad. Teste sur un scalaire et un tableau.


```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_grad(z):
    s = sigmoid(z)
    return s * (1 - s)

# test rapide
print(sigmoid(0))
print(sigmoid_grad(0))
print(sigmoid(np.array([-1.,0.,1.])))
```

## Exercice 2 — Modifier le MLP pour ajouter L2 regularization
Implémente une version simple de la mise à jour avec pénalité L2 (weight decay) et compare la loss sur validation.


```python
# Solution (exemple) : ajouter lambda_reg * W dans le gradient
def add_l2(grads, model, lambda_reg=1e-3):
    grads['W1'] += lambda_reg * model.W1
    grads['W2'] += lambda_reg * model.W2
    return grads

# utilisation dans la boucle d'entraînement (ex)
# grads = model.backward(...); grads = add_l2(grads, model); model.update(grads, lr=0.01)
```
